```python
import pandas as pd

df_1 = pd.read_parquet("parquets/BagofWordsPorterStemmerQueriesStemmed.parquet")
df_2 = pd.read_parquet("parquets/BagofWordsLemmatization.parquet")
df_3 = pd.read_parquet("parquets/TF-IDFPorterStemmerQueriesStemmed.parquet")
df_4 = pd.read_parquet("parquets/TF-IDFLemmatization.parquet")
df_5 = pd.read_parquet("parquets/BM25PorterStemmerQueriesStemmed.parquet")
df_6 = pd.read_parquet("parquets/BM25Lemmatization.parquet")
df_7 = pd.read_parquet("parquets/BM25-LemmatizationNumbersReplaced.parquet")
df_8 = pd.read_parquet("parquets/BM25LemmatizationNumbersDeleted.parquet")
df_9 = pd.read_parquet("parquets/BM25LemmatizationCrossEncoder.parquet")
df_10 = pd.read_parquet("parquets/BM25LemmatizationCrossEncoderWithQueryExpansion.parquet")
df_11 = pd.read_parquet("parquets/BM25PseudoRelevanceFeedback.parquet")
```

```python
df = pd.concat([df_1, df_2, df_3, df_4, df_5, df_6, df_7, df_8, df_9, df_10, df_11], ignore_index=True)
df
```

```python
df["f_score_5_mean"]
```

```python
df_diff = pd.DataFrame(
    {
        "Difference": "Reranking Improvement"
    },
    index=[0]
)
```

```python
df_diff
```

```python
for column in df.columns:
    if column != "Method":
        column_name = column + "_improvement"

        if df.loc[8, column] != 0:            
            df_diff[column_name] = 100 * ((df.loc[9, column] - df.loc[8, column]) / df.loc[8, column])
        else:
            df_diff[column_name] = 0.0
```

```python
df_diff
```

```python
df_diff.columns
```

```python
metrics_dict = {
    "metric": [
        "recall_5_mean_improvement",
        "recall_10_mean_improvement",
        "precision_5_mean_improvement",
        "precision_10_mean_improvement",
        "f_score_5_mean_improvement",
        "f_score_10_mean_improvement",
        "MAP_5_improvement",
        "MAP_10_improvement",
        "NDCG_5_mean_improvement",
        "NDCG_10_mean_improvement"
    ],
    "value": [
        df_diff['recall_5_mean_improvement'].apply(lambda x: f"{x:.8f}")[0],
        df_diff['recall_10_mean_improvement'].apply(lambda x: f"{x:.8f}")[0],
        df_diff['precision_5_mean_improvement'].apply(lambda x: f"{x:.8f}")[0],
        df_diff['precision_10_mean_improvement'].apply(lambda x: f"{x:.8f}")[0],
        df_diff['f_score_5_mean_improvement'].apply(lambda x: f"{x:.8f}")[0],
        df_diff['f_score_10_mean_improvement'].apply(lambda x: f"{x:.8f}")[0],
        df_diff['MAP_5_improvement'].apply(lambda x: f"{x:.8f}")[0],
        df_diff['MAP_10_improvement'].apply(lambda x: f"{x:.8f}")[0],
        df_diff['NDCG_5_mean_improvement'].apply(lambda x: f"{x:.8f}")[0],
        df_diff['NDCG_10_mean_improvement'].apply(lambda x: f"{x:.8f}")[0]
    ]
}

df_metric = pd.DataFrame(metrics_dict)
df_metric
```
