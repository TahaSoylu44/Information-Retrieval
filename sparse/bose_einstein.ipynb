```python
import numpy as np

def bose_einstein_weight(word: str, most_doc_list: list) -> float:  # bir kelimenin bo1 weight değerini veren fonksiyon
    tf = 0

    for mytuple in most_doc_list:
        doc_id = mytuple[1]
        doc_index = doc_id_vs_index[doc_id]
        word_column = count_vect.vocabulary_.get(word)

        if word_column is not None:
            tf += int(bag_of_words[doc_index, word_column])
        else:
            print("WARNING")
            print(f"{word} dokümanlar içerisinde bulunamadı!")
            print("Exiting...")
            return

    F = int(np.sum(bag_of_words[:, word_column]))
    Q = F / 369721  # 369721 toplam doküman sayısı

    if Q == 0:
        return 0
    else:
        return tf * np.log2((1 + Q) / Q) + np.log2(1 + Q)
```

```python
doc_id_vs_index = defaultdict(int)

for key, value in doc_dict.items():
    doc_id_vs_index[value] = key
```

```python
import heapq
from collections import defaultdict

def get_expanded_queries(num_of_doc: int, num_of_word: int) -> list:
    """bo1 tekniği ile genişletilmiş sorgular"""
    docs_with_scores = defaultdict(list)    # her sorgu için num_of_doc tane top doküman biriktirilir.

    for query_id in query_ids:
        for i in range(num_of_doc):
            if i < len(query_id_and_its_doc_tuples[query_id]):
                docs_with_scores[query_id].append((query_id_and_its_doc_tuples[query_id][i][0], 
                                                query_id_and_its_doc_tuples[query_id][i][1]))

    feature_names = count_vect.get_feature_names_out()
    
    expanded_queries = []

    for idx in range(len(query_ids)):
        query_list = lemmatized_tokenized_queries[idx].copy()
        query_list = query_list * 2 # orijinal query ağırlığı daha fazla olmalı

        doc_list = docs_with_scores[query_ids[idx]]
        
        unique_word_columns = set()
        for mytuple in doc_list:
            doc_id = mytuple[1]
            doc_index = doc_id_vs_index[doc_id]
            unique_word_columns.update(bag_of_words[doc_index].indices)

        pq = [] # Priority Queue ile bo1 weight değeri yüksek max num_of_word tane word tutacağım.
        # Default Min Heap

        for column in unique_word_columns:
            word = feature_names[column]
            weight = bose_einstein_weight(word, doc_list)
            heapq.heappush(pq, (weight, word))

            if len(pq) > num_of_word:   # sadece en büyük num_of_word tane kelime tutulacak
                heapq.heappop(pq)

        while pq:
            best_word = heapq.heappop(pq)[1]
            query_list.append(best_word)    # her sorgu için genişletilmiş query

        expanded_queries.append(query_list)

    return expanded_queries
```

```python
from sklearn.metrics import ndcg_score

def get_ndcg(similarities: np.array, value: int) -> float:
    doc_id_score_dict = defaultdict(float)  # key: doc_id   value: its score
    ndcg_value_list = []

    for idx in range(len(query_ids)):
        query_id = query_ids[idx]
        scoredoc_object_list = score_doc_dict[query_id]

        for scoredoc_object in scoredoc_object_list:
            doc_id_score_dict[scoredoc_object.doc_id] = scoredoc_object.score

        similarity = similarities[idx]
        index_of_found_docs = np.where(similarity > 0)[0]
        y_true, y_score = [], []

        for index in index_of_found_docs:
            y_score.append(similarity[index])
            y_true.append(doc_id_score_dict.get(doc_id[index], 0.0))

        if len(y_score) == 0:
            ndcg_value_list.append(0.0)
            continue

        if len(y_score) == 1:   # sklearn ndcg fonksiyonu çalışabilemsi için 1 den fazla eleman gerek, eğer boyut 1 ise her iki listeye dummy değişkenler ekledim.
            y_true.append(0.0)
            y_score.append(0.0)

        if len(y_true) == len(y_score):
            ndcg_value = ndcg_score(np.asarray([y_true]), np.asarray([y_score]), k=value)

            try:    # bazen NDCG None olabiliyor, listeye None değer ekleyemem
                trial = int(ndcg_value) # eğer bu kısım patlamıyorsa NDCG value düzgün bir float değerdir.
                ndcg_value_list.append(ndcg_value)  # ekle gitsin
            except ValueError:
                ndcg_value_list.append(0.0) # None geliyorsa, bir yerde "ZeroDivison" gelmiş olabilir, 0 ekledim.
        else:
            print("There is a problem with query", query_id)

    return np.mean(ndcg_value_list)
```

```python
def get_mean_ndcg(query_list: list) -> float:
    scores = []

    for query in query_list:
        scores.append(bm25.get_scores(query))

    ndcg_5_mean = get_ndcg(scores, 5)   # top 5 elemente bakarak ndcg hesapla
    ndcg_10_mean = get_ndcg(scores, 10) # top 5 elemente bakarak ndcg hesapla

    return (ndcg_5_mean + ndcg_10_mean) / 2 # amacım her ikisini de yüksek tutmak olduğu için ortalamasını aldım.
```

```python
import optuna

def objective(trial):
    num_of_doc = trial.suggest_int('num_of_doc', 1, 5)
    num_of_word = trial.suggest_int('num_of_word', 1, 3)

    query_list = get_expanded_queries(num_of_doc, num_of_word)
    return get_mean_ndcg(query_list)    # ndcg değeri maximize edilecek değer
```
