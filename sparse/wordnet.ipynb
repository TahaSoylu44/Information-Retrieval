```python
from nltk.corpus import wordnet as wn

synsets = wn.synsets("car")
synsets
```

```python
car = synsets[0]
print("Name:", car.name())
print("Definition:", car.definition())
print("Examples:", car.examples())
```

```python
car.lemma_names()
```

```python
type(car.lemma_names())
```

```python
dog = wn.synset("dog.n.01")
cat = wn.synset("cat.n.01")
```

```python
similarity = dog.path_similarity(cat)
print(f"Path similarity between dog and cat: {similarity}")
```

```python
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

seq = "I love eating apple"
key_word = "apple"

temp = lesk(word_tokenize(seq), key_word)
```
